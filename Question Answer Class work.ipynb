```python
1. Function to print “Hello, World!”

def hello_world():
    print("Hello, World!")
```

```python
2. Function to add two numbers

def add_numbers(a, b):
    return a + b
```

```python
3. Function to find the square of a number

def square(num):
    return num * num
```

```python
4. Function to find the factorial of a number

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
```

```python
5. Function to check if a number is even or odd

def even_or_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

```python
6. Function to find the maximum of three numbers

def max_of_three(a, b, c):
    return max(a, b, c)
```

```python
7. Function to check whether a number is prime

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
8. Function to calculate the sum of first n natural numbers

def sum_natural(n):
    return n * (n + 1) // 2
```

```python
9. Function to return the length of a string

def string_length(s):
    return len(s)
```

```python
10. Function to reverse a string

def reverse_string(s):
    return s[::-1]
```
