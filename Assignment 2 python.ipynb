```python
for i in range(1, 51):
    if i > 40:
        break
    if i % 10 == 0:
        continue
    if i % 2 == 0:
        print(i)
```

```python
num = int(input("Enter a number: "))
temp = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10

if temp == rev:
    print("Palindrome Number")
else:
    print("Not a Palindrome")
```

```python
n = int(input("Enter number of terms: "))

a, b = 0, 1
print("Fibonacci Series:")

for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

```python
num = int(input("Enter a number: "))
temp = num
order = len(str(num))
sum = 0

while num > 0:
    digit = num % 10
    sum += digit ** order
    num //= 10

if sum == temp:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
```
