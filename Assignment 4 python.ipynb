```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("iris")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['sepal_length'].hist(bins=20)
plt.title("Histogram of Sepal Length")
plt.show()
```

```python
sns.kdeplot(df['petal_length'], fill=True)
plt.title("KDE Plot of Petal Length")
plt.show()
```

```python
sns.countplot(x='species', data=df)
plt.title("Count of Each Species")
plt.show()
```

```python
sns.scatterplot(x='sepal_length', y='petal_length', hue='species', data=df)
plt.title("Sepal Length vs Petal Length")
plt.show()
```

```python
sns.lineplot(x=df.index, y='sepal_width', data=df)
plt.title("Line Plot of Sepal Width")
plt.show()
```

```python
sns.boxplot(x='species', y='petal_width', data=df)
plt.title("Box Plot of Petal Width by Species")
plt.show()
```

```python
sns.violinplot(x='species', y='sepal_length', data=df)
plt.title("Violin Plot of Sepal Length")
plt.show()
```

```python
sns.pairplot(df, hue='species')
plt.show()
```

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
